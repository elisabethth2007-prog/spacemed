```python
if planets[p] >1:
    print(f"{p} has a larger mass than the Earth")
elif planets [p] <1:
    print(f"{p} has a smaller mass than the Earth")
else:
    print(f"{p} has the same mass as the Earth")
```

```python
for p in planets:
    if planets[p] >1:
        print(f"{p} has a larger mass than the Earth")
    elif planets [p] <1:
        print(f"{p} has a smaller mass than the Earth")
    else:
        print(f"{p} has the same mass as the Earth")
```
