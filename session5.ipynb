```python
precip = pandas.read_csv("data/produkt_precipitation_399_akt.txt",
                             encoding="iso-8859-1", delimiter=" *; *",
                                 engine="python")
```

```python
precip.head()
```

```python
precip.MESS_DATUM = pandas.to_datetime(precip.MESS_DATUM, format="%Y%m%d%H")
```

```python
precip['date_time'] = pandas.to_datetime(precip.MESS_DATUM, format="%Y%m%d%H") #create new column
```

```python
precip_alex = precip[['date_time', "NIEDERSCHLAGSHOEHE"]]
```

```python
precip_alex = precip_alex.rename(columns={"date_time":"date",
                                          "NIEDERSCHLAGSHOEHE":"precipitation"})
```

```python
precip_alex = precip_alex.set_index("date")
```

```python
precip_alex
```

```python
monthly = precip_alex.resample("ME").sum() #ME is months end
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 1, figsize=(10, 6))

# Top plot
monthly.plot(ax=axes[0]) 
axes[0].set_title("Monthly Data")

# Bottom plot
monthly.anomaly.plot(ax=axes[1])
axes[1].set_title("Anomaly")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

#axes is a list/array of subplot objects
fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True) #2 rows 1 column, share the x axis

# Top plot
monthly[['precipitation', 'climate']].plot (ax=axes[0]) #ax=axes[0]:“Don’t create a new plot—draw it on the first axis in axes.”
axes[0].set_title("Monthly Data")

# Bottom plot
monthly.anomaly.plot(ax=axes[1])
axes[1].set_title("Anomaly")

plt.tight_layout()
plt.show()
```

Combine all three plots:

```python
#axes is a list/array of subplot objects
fig, axes = plt.subplots(3, 1, figsize=(10, 6), sharex=True) #3 rows 1 column, share the x axis

# Top plot
monthly[['precipitation', 'climate']].plot (ax=axes[0]) #ax=axes[0]:“Don’t create a new plot—draw it on the first axis in axes.”
axes[0].set_title("Monthly Data")
axes[0].set_ylabel("Precipitation [mm]")

# Middle plot
x = monthly.index
monprec = monthly['precipitation']
monclim = monthly['climate']

monthly[['precipitation', 'climate']].plot(ax=axes[1])

axes[1].fill_between(x, monprec, monclim, where=(monprec > monclim), color='blue', alpha=0.3)
axes[1].fill_between(x, monprec, monclim, where=(monprec <= monclim), color='orange', alpha=0.3)

axes[1].set_title("Monthly Data")
axes[1].set_ylabel("Precipitation [mm]")

# Bottom plot
monthly.anomaly.plot(ax=axes[2])
axes[2].axhline(0)
axes[2].set_title("Anomaly")
axes[2].set_ylabel("Anomaly [mm]")


plt.tight_layout()
plt.show()
```
